```python
MODEL_ID = "m-f6812ff886404598946420b8d3ea713a"
MODEL_URI = f"models:/{MODEL_ID}"
```

```python
%sql
SELECT 'silver_listings' AS table_name, city, COUNT(*) AS rows
FROM airbnb_lakehouse.silver_listings
GROUP BY city

UNION ALL

SELECT 'gold_listings_features', city, COUNT(*)
FROM airbnb_lakehouse.gold_listings_features
GROUP BY city

UNION ALL

SELECT 'gold_city_metrics', city, COUNT(*)
FROM airbnb_lakehouse.gold_city_metrics
GROUP BY city

UNION ALL

SELECT 'gold_model_predictions', city, COUNT(*)
FROM airbnb_lakehouse.gold_model_predictions
GROUP BY city

ORDER BY table_name, city;
```

```python
%sql
DESCRIBE HISTORY airbnb_lakehouse.silver_listings;
```

```python
%sql
DESCRIBE HISTORY airbnb_lakehouse.gold_listings_features;
```

```python
%sql
DESCRIBE HISTORY airbnb_lakehouse.gold_city_metrics;
```

```python
tables = [
    "silver_listings",
    "gold_listings_features",
    "gold_city_metrics",
]

for name in tables:
    spark.sql(f"""
        CREATE TABLE IF NOT EXISTS
            airbnb_lakehouse.backup_pre_montreal_{name}
        AS SELECT *
        FROM airbnb_lakehouse.{name}
    """)
```
