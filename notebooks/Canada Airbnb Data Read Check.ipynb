# Canada Airbnb — Data Read Check

Verifies that the raw Airbnb data files (listings, calendar, reviews) for Montreal, Toronto, and Vancouver in `/Shared/canada-airbnb/data` can be read, and shows a sample of each.

```python
import pandas as pd

CITIES = ["montreal", "toronto", "vancouver"]
FILES = ["listings.csv.gz", "calendar.csv.gz", "reviews.csv.gz"]

read_results = []
for city in CITIES:
    for fname in FILES:
        path = f"/Workspace/Shared/canada-airbnb/data/{city}/{fname}"
        try:
            sample = pd.read_csv(path, compression="gzip", nrows=5)
            read_results.append({
                "city": city,
                "file": fname,
                "status": "OK",
                "columns": len(sample.columns),
                "sample_rows": len(sample),
            })
        except Exception as e:
            read_results.append({
                "city": city,
                "file": fname,
                "status": f"FAILED: {type(e).__name__}: {e}",
                "columns": None,
                "sample_rows": 0,
            })

display(pd.DataFrame(read_results))
```
